# Bronze Layer

Esta capa almacena los datos crudos provenientes
del dataset Wanderbricks sin transformaciones,
siguiendo la arquitectura Medallion.

```python
%sql
CREATE SCHEMA IF NOT EXISTS bronze;

CREATE SCHEMA IF NOT EXISTS silver;

CREATE SCHEMA IF NOT EXISTS gold;
```

## Creación de tabla bronze_bookings

Se copia la tabla original bookings
desde el catálogo samples.wanderbricks
hacia la capa Bronze.

```python
%sql
CREATE TABLE bronze.bronze_bookings AS
SELECT *
FROM samples.wanderbricks.bookings;
```

## Validación de Datos


```python
%sql
SELECT *
FROM bronze.bronze_bookings
LIMIT 10;
```
